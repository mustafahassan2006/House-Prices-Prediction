According to data rules Na dosent mean data missing it means the feature is not present in that house so replacing Nan with none

```python
no_feature_cols = [
    'PoolQC', 'MiscFeature', 'Alley', 'Fence', 'FireplaceQu',
    'GarageType', 'GarageFinish', 'GarageQual', 'GarageCond',
    'BsmtQual', 'BsmtCond', 'BsmtExposure', 'BsmtFinType1', 'BsmtFinType2','MasVnrType'
]

Train_df[no_feature_cols] = Train_df[no_feature_cols].fillna('None')
Train_df['MasVnrArea'] = Train_df['MasVnrArea'].fillna(0)
Test_df[no_feature_cols] = Train_df[no_feature_cols].fillna('None')
Test_df['MasVnrArea'] = Train_df['MasVnrArea'].fillna(0)
```

```python
Train_df.info()
```

Still have some null because looking at the data those values are actualy missing like Electrical and LotFrontage    

```python
Train_df['GarageYrBlt']=Train_df['GarageYrBlt'].fillna('YearBuilt')
Test_df['GarageYrBlt']=Test_df['GarageYrBlt'].fillna('YearBuilt')
```

```python
from sklearn.model_selection import train_test_split
Train_Split_df, Val_df=train_test_split(Train_df, test_size=0.2, random_state=40)
```

```python
Target_col=Train_Split_df['SalePrice']
Input_col=Train_Split_df.drop(columns=['Id', 'SalePrice', 'Utilities'])
Val_Target=Val_df['SalePrice']
Val_df_Input=Val_df.drop(columns=['Id', 'SalePrice', 'Utilities'])
Test_df_input=Test_df.drop(columns=['Id', 'Utilities'])
```

```python
null_counts = Input_col.isnull().sum()
null_summary = null_counts[null_counts > 0].sort_values(ascending=False)
print(null_summary)
```

```python
from sklearn.impute import SimpleImputer

Imputer_median=SimpleImputer(strategy='median')
Imputer_frq=SimpleImputer(strategy='most_frequent')
Imputer_median.fit(Input_col[['LotFrontage']])
Imputer_frq.fit(Input_col[['Electrical']])

```

```python
Input_col['LotFrontage']=Imputer_median.transform(Input_col[['LotFrontage']]).ravel()
Input_col['Electrical']=Imputer_frq.transform(Input_col[['Electrical']]).ravel()
Val_df_Input['LotFrontage']=Imputer_median.transform(Val_df_Input[['LotFrontage']]).ravel()
Val_df_Input['Electrical']=Imputer_frq.transform(Val_df_Input[['Electrical']]).ravel()
Test_df_input['LotFrontage']=Imputer_median.transform(Test_df_input[['LotFrontage']]).ravel()
Test_df_input['Electrical']=Imputer_frq.transform(Test_df_input[['Electrical']]).ravel()
```

Encodding catagorical data

```python
Input_col['GarageYrBlt'] = pd.to_numeric(Input_col['GarageYrBlt'], errors='coerce')
Input_col['GarageYrBlt'] = Input_col['GarageYrBlt'].fillna(0)
Val_df_Input['GarageYrBlt'] = pd.to_numeric(Input_col['GarageYrBlt'], errors='coerce')
Val_df_Input['GarageYrBlt'] = Input_col['GarageYrBlt'].fillna(0)
Test_df_input['GarageYrBlt'] = pd.to_numeric(Input_col['GarageYrBlt'], errors='coerce')
Test_df_input['GarageYrBlt'] = Input_col['GarageYrBlt'].fillna(0)
```

```python
numeric_colls=Input_col.select_dtypes(include=['number']).columns.tolist()
categorical_cols = Input_col.select_dtypes(exclude=['number']).columns.tolist()
```

```python
Input_col
```

```python
from sklearn.preprocessing import StandardScaler
Scaler=StandardScaler()
Scaler.fit(Input_col[numeric_colls])
Input_col[numeric_colls]=Scaler.transform(Input_col[numeric_colls])
Val_df_Input[numeric_colls]=Scaler.transform(Val_df_Input[numeric_colls])
Test_df_input[numeric_colls]=Scaler.transform(Test_df_input[numeric_colls])
```

```python
Input_col
```

```python
Input_col[categorical_cols].nunique()
```

Maping

```python
Input_col['CentralAir']=Input_col['CentralAir'].map({'Y':1,'N':0})
Input_col['Street']=Input_col['Street'].map({'Pave':1,'Grvl':0})
Input_col['PavedDrive'] = Input_col['PavedDrive'].map({'Y': 3, 'P': 2, 'N': 1}).fillna(0)
Input_col['Fence']=Input_col['Fence'].map({'GdPrv':4,'MnPrv':3, 'GdWo':2, 'MnWw':1, 'None':0})
#
Val_df_Input['CentralAir']=Val_df_Input['CentralAir'].map({'Y':1,'N':0})
Val_df_Input['Street']=Val_df_Input['Street'].map({'Pave':1,'Grvl':0})
Val_df_Input['PavedDrive'] = Val_df_Input['PavedDrive'].map({'Y': 3, 'P': 2, 'N': 1}).fillna(0)
Val_df_Input['Fence']=Val_df_Input['Fence'].map({'GdPrv':4,'MnPrv':3, 'GdWo':2, 'MnWw':1, 'None':0})
#
Test_df_input['CentralAir']=Test_df_input['CentralAir'].map({'Y':1,'N':0})
Test_df_input['Street']=Test_df_input['Street'].map({'Pave':1,'Grvl':0})
Test_df_input['PavedDrive'] = Test_df_input['PavedDrive'].map({'Y': 3, 'P': 2, 'N': 1}).fillna(0)
Test_df_input['Fence']=Test_df_input['Fence'].map({'GdPrv':4,'MnPrv':3, 'GdWo':2, 'MnWw':1, 'None':0})
#make a list of all quality columns 
Qual_col=Input_col.columns[Input_col.columns.str.contains('Qual$|Cond$|QC$|Qu$')].tolist()

qual_map = {'Ex': 5, 'Gd': 4, 'TA': 3, 'Fa': 2, 'Po': 1, 'None': 0}
for col in Qual_col:
    Input_col[col]=Input_col[col].map(qual_map).fillna(0)
    Val_df_Input[col]=Val_df_Input[col].map(qual_map).fillna(0)
    Test_df_input[col]=Test_df_input[col].map(qual_map).fillna(0)
    
```

```python
Not_In_Qual_Col_But_Qual=['CentralAir','Street','PavedDrive','Fence']
All_qual = list(set(Qual_col+Not_In_Qual_Col_But_Qual))
```

```python
Hot_Encoding_Cols=list(set(categorical_cols)-set(All_qual))
```

```python
Hot_Encoding_Cols
```

One Hot Encodding

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
encoder.fit(Input_col[Hot_Encoding_Cols])
```

```python
encoder.categories_
```

```python
encoder_col=list(encoder.get_feature_names_out(Hot_Encoding_Cols))
print(encoder_col)
```

```python
Input_col[encoder_col]=encoder.transform(Input_col[Hot_Encoding_Cols])
Val_df_Input[encoder_col]=encoder.transform(Val_df_Input[Hot_Encoding_Cols])
Test_df_input[encoder_col]=encoder.transform(Test_df_input[Hot_Encoding_Cols])
```

```python
Input_col
```

```python
raw_cat_cols = [
    'MSZoning', 'Alley', 'LotShape', 'LandContour', 'LotConfig', 'LandSlope', 
    'Neighborhood', 'Condition1', 'Condition2', 'BldgType', 'HouseStyle', 
    'RoofStyle', 'RoofMatl', 'Exterior1st', 'Exterior2nd', 'MasVnrType', 
    'Foundation', 'BsmtExposure', 'BsmtFinType1', 'BsmtFinType2', 'Heating', 
    'Functional', 'GarageType', 'GarageFinish', 'MiscFeature', 'SaleType', 
    'SaleCondition','Electrical'
]
Input_col     = Input_col.drop(columns=[col for col in raw_cat_cols if col in Input_col.columns])
Val_df_Input  = Val_df_Input.drop(columns=[col for col in raw_cat_cols if col in Val_df_Input.columns])
Test_df_input = Test_df_input.drop(columns=[col for col in raw_cat_cols if col in Test_df_input.columns])
```

```python
Input_col
```

```python
from sklearn.linear_model import Ridge
```

```python
model = Ridge(alpha=10.0)
model.fit(Input_col, Target_col)
```

```python
Val_df_Input['GarageYrBlt']  = Val_df_Input['GarageYrBlt'].fillna(Val_df_Input['YearBuilt'])
```

```python
predictio_val=model.predict(Val_df_Input)
```

```python
Prediction
```

```python
import numpy as np
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# 1. Generate predictions on validation set
val_preds = model.predict(Val_df_Input)

# Safety clip: Ensure no price is below 0
val_preds = np.maximum(0, val_preds)

# 2. Calculate key metrics
rmse = np.sqrt(mean_squared_error(Val_Target, val_preds))
mae  = mean_absolute_error(Val_Target, val_preds)
r2   = r2_score(Val_Target, val_preds)

print(f"Validation RMSE: ${rmse:,.2f}")
print(f"Validation MAE:  ${mae:,.2f}")
print(f"Validation R²:   {r2:.4f}")
```
